# Venezuela Economic Analysis (1960-2023)

**Objective:**  
Analyze how Venezuela’s dependence on natural resources, especially oil, influenced its economic performance over the past 60+ years.  

This notebook demonstrates:
- Data cleaning and manipulation using **pandas**
- Trend analysis and visualizations using **matplotlib**
- Deriving insights that show **economic patterns and value creation**

```python
df = df.sort_values('year')
df.head()
```

```python
df = df.rename(columns={'year': 'Year'})
# Ensure Year is a number #data
df['Year'] = pd.to_numeric(df['Year'], errors='coerce')
# Now we can safely sort
df = df.sort_values('Year')
df.head(2)
```

```python
df.columns
```

```python
df['gdp_per_oil_rent'] = df['gdp_current_usd'] / df['oil_rents_pct_gdp']
```

```python
# 5-year rolling average for GDP
df['gdp_rolling'] = df['gdp_current_usd'].rolling(window=5).mean()

# 5-year rolling average for Oil Rents
df['oil_rents_rolling'] = df['oil_rents_pct_gdp'].rolling(window=5).mean()
```

### Data Manipulation 

- **GDP per unit of Oil Rent:** Measures economic efficiency relative to resource dependency.  
- **Rolling averages (5-year):** Smooth volatility to better visualize long-term trends in GDP and oil dependency.


```python


plt.figure(figsize=(14,7))
plt.plot(df['Year'], df['gdp_rolling'], label='GDP (5-year avg)', color='blue', linewidth=2)
plt.plot(df['Year'], df['oil_rents_rolling'], label='Oil Rents (% of GDP, 5-year avg)', color='green', linewidth=2)
plt.title("Venezuela GDP vs Oil Dependency (1960-2023, 5-Year Rolling Average)", fontsize=16)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Value", fontsize=12)
plt.legend()
plt.grid(True)
plt.show()
```

### Insights

- GDP closely follows trends in oil dependency, showing Venezuela's vulnerability to oil market fluctuations.  
- Rolling averages make long-term patterns visible, highlighting periods of growth and contraction.  
- This chart demonstrates the strong correlation between resource dependency and economic performance.


```python
# Calculate year-over-year % change in GDP
df['gdp_pct_change'] = df['gdp_current_usd'].pct_change() * 100

# Identify sharp drops (more than 10%)
gdp_shocks = df[df['gdp_pct_change'] < -10]

# Display these years
gdp_shocks[['Year', 'gdp_current_usd', 'gdp_pct_change']]
```

### Economic Shocks (Large GDP Drops)

The table above identifies years with GDP declines greater than 10%, highlighting periods of economic stress.  
These declines often coincide with oil price shocks or political events, demonstrating Venezuela’s vulnerability to resource dependency.


```python
# Top 5 years by GDP per unit of Oil Rent
efficiency_peaks = df[['Year', 'gdp_current_usd', 'oil_rents_pct_gdp', 'gdp_per_oil_rent']].sort_values(
    by='gdp_per_oil_rent', ascending=False).head()

efficiency_peaks
```

### Economic Efficiency Peaks

By analyzing GDP per unit of oil revenue, we can see which years the economy generated the most value relative to oil dependency.  
This helps identify periods of better economic management and efficiency.


```python
plt.figure(figsize=(14,7))
plt.scatter(df['oil_rents_pct_gdp'], df['gdp_growth_pct'], color='red')
plt.title("GDP Growth vs Oil Dependency (%)", fontsize=16)
plt.xlabel("Oil Rents (% of GDP)", fontsize=12)
plt.ylabel("GDP Growth (%)", fontsize=12)
plt.grid(True)
plt.show()
```

### Insight

- The scatter plot shows that very high oil dependency is often associated with **volatile GDP growth**.  
- Periods of moderate dependency tend to have **more stable growth**, highlighting the risks of over-reliance on natural resources.


# Summary of Key Insights

1. Venezuela's GDP trends closely follow oil dependency, illustrating high resource reliance.  
2. Several years experienced sharp GDP drops (>10%), highlighting vulnerability to oil shocks and political crises.  
3. Rolling averages and derived metrics (GDP per Oil Rent) provide a clearer view of long-term trends and efficiency.  
4. Scatter analysis shows that extremely high resource dependency often leads to economic volatility, emphasizing the need for diversification.  
5. These analyses demonstrate **how data-driven insights can inform economic planning and policy decisions**.
